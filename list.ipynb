# `list` class

If you want to store more than one value you can use a `list` class. In order to use a list in python, you have to enclose the 
values in square brackets `[ ]`.

Link: [Documentation of `lst` class](https://docs.python.org/3.13/library/stdtypes.html#list)

```python
[1, 2, 3, 2.0, "data"]
```

```python
type([1, 2, 3, 2.0, "data"])
```

```python
[1, 2, 3] + [3]
```
